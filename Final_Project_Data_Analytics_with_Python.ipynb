```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel('Sleep_health_and_lifestyle_dataset.xlsx', header= 1)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe().round(2)
```

```python
# no duplicates in the dataset

df.duplicated().sum()
```

```python
# remove rows that are missing data

df.isnull().sum()
df = df.dropna()
df.isnull().sum()

# rows were removed to clean up the data
```

```python
# visualization after removing null values


sns.barplot(data=df, x="Occupation", y="Sleep Duration")

plt.title("Average Sleep Duration by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Sleep Duration (Hours)")
plt.xticks(rotation=90)
plt.show()
```

```python
# visualization of data after removing rows with missing values


sns.histplot(data=df, x="Daily Steps")
plt.title("Distribution of Daily Steps")
plt.xlabel("Daily Steps")
plt.ylabel("Amount of people")
plt.show()
```

```python
# data comparing Daily Steps and Blood Pressure Level


sns.scatterplot(data=df, x="Daily Steps", y="Blood Pressure")
plt.title("Daily Steps vs. Blood Pressure Level")
plt.xlabel("Daily Steps")
plt.ylabel("Blood Pressure Level")
plt.show()

```

```python
# using correlation analysis of the data with numerical values to see the level of correlation

print(df.describe().round(2))

df.corr(numeric_only=True)
```
